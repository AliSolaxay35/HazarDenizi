```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os
from google.colab import files
```

```python
os.makedirs('images', exist_ok=True)

uploaded = files.upload()

for filename in uploaded.keys():
  dest = os.path.join('images', filename)
  os.rename(filename, dest)
  print(f'{filename} Is Successfully Uploaded To {dest}')

print("\nContents Of Images Folder:")
print(os.listdir('images'))
```

```python
image = cv2.imread('images/input.jpg')
height, width, _ = image.shape

T = np.float32([[1, 0, 100], [0, 1, 30]])

img_translation = cv2.warpAffine(image, T, (width + 100, height + 30))

plt.imshow(img_translation[..., ::-1])
plt.axis('off')
plt.title('Translation')

plt.show()

print(T)
```
