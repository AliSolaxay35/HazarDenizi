```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os
from google.colab import files
```

```python
os.makedirs('images', exist_ok=True)

uploaded = files.upload()

for filename in uploaded.keys():
    dest = os.path.join('images', filename)
    os.rename(filename, dest)
    print(f"File '{filename}' Moved To '{dest}'.")

print("\nContents Of Images Folder:")
print(os.listdir('images'))
```

```python
rotation_amount = -20
theta = rotation_amount * np.pi / 180.0

image = cv2.imread('images/input.jpg')
height, width, _ = image.shape

T = np.float32([[np.cos(theta), -np.sin(theta), 0], [np.sin(theta), np.cos(theta), 0]])
img_translation = cv2.warpAffine(image, T, (width, height))

plt.imshow(img_translation[..., ::-1])
plt.axis('off')

plt.show()
```

```python
rotation_amount = 10

theta = rotation_amount * np.pi / 180.0

image = cv2.imread('images/input.jpg')
height, width, _ = image.shape

T = np.float32([[np.cos(theta), -np.sin(theta), 0], [np.sin(theta), np.cos(theta), -50]])
img_translation = cv2.warpAffine(image, T, (width, height))

plt.imshow(img_translation[..., ::-1])
plt.axis('off')

plt.show()
```

```python
image = cv2.imread('images/input.jpg')
height, width = image.shape[:2]

rotation_matrix = cv2.getRotationMatrix2D((width / 2, height / 2), 20, 5)
rotated_image = cv2.warpAffine(image, rotation_matrix, (width, height))

plt.subplot(1, 2, 1)
plt.imshow(image[..., ::-1])
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(rotated_image[..., ::-1])
plt.title('Rotated Image')
plt.axis('off')

plt.show()
```

```python
image = cv2.imread('images/input.jpg')

transposed = cv2.transpose(image)

plt.subplot(1, 2, 1)
plt.imshow(image[..., ::-1])
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(transposed[..., ::-1])
plt.title('Transposed Image')
plt.axis('off')

plt.show()
```

```python
flipped = cv2.flip(image, 1)

plt.subplot(1, 2, 1)
plt.imshow(image[..., ::-1])
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(flipped[..., ::-1])
plt.title('Flipped Image')
plt.axis('off')

plt.show()
```

```python
flipped = cv2.flip(image, 0)

plt.subplot(1, 2, 1)
plt.imshow(image[..., ::-1])
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(flipped[..., ::-1])
plt.title('Flipped Image')
plt.axis('off')

plt.show()
```
