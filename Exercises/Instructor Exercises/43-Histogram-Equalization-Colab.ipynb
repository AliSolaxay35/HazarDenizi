```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os
from google.colab import files
```

```python
os.makedirs('images', exist_ok=True)

uploaded = files.upload()

for filename in uploaded.keys():
  dest = os.path.join('images', filename)
  os.rename(filename, dest)
  print(f"File '{filename}' Moved To '{dest}'.")

print("\nContents Of Images Folder:")
print(os.listdir('images'))
```

```python
image = cv2.imread('images/xray.jpg', 0)
result = cv2.equalizeHist(image)

plt.figure(figsize=[8, 4])

plt.subplot(1, 2, 1)
plt.imshow(image, cmap='gray')
plt.title('Original Image')

plt.subplot(1, 2, 2)
plt.imshow(result, cmap='gray')
plt.title('Equalized Image')

plt.show()
```

```python
image = cv2.imread('images/mid.png')

channels = cv2.split(image)
eq_channels = []

for channel in channels:
  eq_channels.append(cv2.equalizeHist(channel))

eq_image = cv2.merge(eq_channels)

plt.figure(figsize=[12, 7])

plt.subplot(2, 2, 1)
plt.imshow(image[..., ::-1])
plt.title('Original Image')

plt.subplot(2, 2, 3)
plt.imshow(eq_image[..., ::-1])
plt.title('Equalized Image')

colors = ('b', 'g', 'r')

plt.subplot(2, 2, 2)
for i, color in enumerate(colors):
  hist = cv2.calcHist([image], [i], None, [256], [0, 256])
  plt.plot(hist, color=color)

plt.subplot(2, 2, 4)
for i, color in enumerate(colors):
  hist = cv2.calcHist([eq_image], [i], None, [256], [0, 256])
  plt.plot(hist, color=color)

plt.show()
```
