```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os
from google.colab import files
```

```python
os.makedirs('images', exist_ok=True)

uploaded = files.upload()

for filename in uploaded.keys():
    dest = os.path.join('images', filename)
    os.rename(filename, dest)
    print(f"File '{filename}' Moved To '{dest}'.")

print("\nContents Of Images Folder:")
print(os.listdir('images'))
```

```python
image = cv2.imread('images/plate.jpg')

kernel1 = np.array([[0, 0, 0], [0, 2, 0], [0, 0, 0]])
kernel2 = np.array([[1, 1, 1], [1, 1, 1], [1, 1, 1]])

final_kernel = kernel1 - kernel2

sharpened_image = cv2.filter2D(image, -1, final_kernel)

plt.figure(figsize=(12, 7))

plt.subplot(1, 2, 1)
plt.imshow(image[..., ::-1])
plt.title('Original Image')

plt.subplot(1, 2, 2)
plt.imshow(sharpened_image[..., ::-1])
plt.title('Sharpened Image')

plt.show()
```

```python
image = cv2.imread('images/plate.jpg')

kernel_sharpening = np.array([[-1, -1, -1], [-1, 9, -1], [-1, -1, -1]])

sharpened_image = cv2.filter2D(image, -1, kernel_sharpening)

plt.figure(figsize=(12, 7))

plt.subplot(1, 2, 1)
plt.imshow(image[..., ::-1])
plt.title('Original Image')

plt.subplot(1, 2, 2)
plt.imshow(sharpened_image[..., ::-1])
plt.title('Sharpened Image')

plt.show()
```
