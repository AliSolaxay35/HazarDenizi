```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os
from google.colab import files
```

```python
os.makedirs('images', exist_ok=True)

uploaded = files.upload()

for filename in uploaded.keys():
  dest = os.path.join('images', filename)
  os.rename(filename, dest)
  print(f"File '{filename}' Moved To '{dest}'.")

print("\nContents Of Images Folder:")
print(os.listdir('images'))
```

```python
img = cv2.imread('images/airplane.jpg', 0)

low_sigma = cv2.GaussianBlur(img, (3, 3), 2)
high_sigma = cv2.GaussianBlur(img, (3, 3), 6)

dog = low_sigma - high_sigma

plt.figure(figsize=(20, 5))

plt.subplot(1, 4, 1)
plt.imshow(img, cmap='gray')
plt.title('Original Image')

plt.subplot(1, 4, 2)
plt.imshow(low_sigma, cmap='gray')
plt.title('Low Sigma')

plt.subplot(1, 4, 3)
plt.imshow(high_sigma, cmap='gray')
plt.title('High Sigma')

plt.subplot(1, 4, 4)
plt.imshow(dog, cmap='gray')
plt.title('Difference of Gaussians')

plt.show()
```
