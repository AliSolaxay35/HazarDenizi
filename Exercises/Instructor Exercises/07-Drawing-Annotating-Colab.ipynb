```python
image = np.zeros((700, 900, 3), np.uint8)
pil_image = Image.fromarray(image)
font_path = "/usr/share/fonts/truetype/liberation/LiberationSans-Regular.ttf"
font = ImageFont.truetype(font_path, 38)
draw = ImageDraw.Draw(pil_image)

arabic_text = get_display(arabic_reshaper.reshape("مرحبا بالعالم"))
draw.text((45, 120), "Merhaba Dunya!", font=font, fill=(100, 170, 0))
draw.text((45, 200), "Hello World!", font=font, fill=(100, 170, 0))
draw.text((45, 280), "Hallo Welt!", font=font, fill=(100, 170, 0))
draw.text((45, 360), arabic_text, font=font, fill=(100, 170, 0))

plt.figure(figsize=(10, 8))
plt.imshow(pil_image)
plt.title("Turkish / English / German / Arabic")
plt.axis('off')
```
