```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
import os
from google.colab import files
from scipy import io
```

```python
os.makedirs('dataset', exist_ok=True)

uploaded = files.upload()

for filename in uploaded.keys():
  dest = os.path.join('dataset', filename)
  os.rename(filename, dest)
  print(f"File '{filename}' Moved To '{dest}'.")

print("\nContents Of Dataset Folder:")
print(os.listdir('dataset'))
```

```python
dataset = io.loadmat('./dataset/Data_hoda_full.mat')
```

```python
x_train_original = np.squeeze(dataset['Data'][:1000])
y_train_original = np.squeeze(dataset['labels'][:1000])
x_test_original = np.squeeze(dataset['Data'][1000:1200])
y_test_original = np.squeeze(dataset['labels'][1000:1200])
```

```python
i = 50

plt.imshow(x_train_original[i], cmap='gray')
plt.title(f'Label: {y_train_original[i]}')

plt.show()
```

```python
x_train_5by5 = [cv2.resize(img, (5, 5)) for img in x_train_original]
x_test_5by5 = [cv2.resize(img, (5, 5)) for img in x_test_original]
```

```python
i = 50

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(x_train_5by5[i], cmap='gray')
plt.title(f'Label: {y_train_original[i]}')

plt.subplot(1, 2, 2)
plt.imshow(x_train_original[i], cmap='gray')
plt.title(f'Label: {y_train_original[i]}')

plt.show()
```
