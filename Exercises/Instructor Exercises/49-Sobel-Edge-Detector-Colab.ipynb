```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os
from google.colab import files
```

```python
os.makedirs('images', exist_ok=True)

print("Please Upload The Required Image:")
print("- little_girl.jpg")

uploaded = files.upload()

for filename in uploaded.keys():
  dest = os.path.join('images', filename)
  os.rename(filename, dest)
  print(f"File '{filename}' Moved To '{dest}'.")

print("\nContents Of Images Folder:")
print(os.listdir('images'))
```

```python
image = cv2.imread('images/little_girl.jpg')

sobel_x = cv2.Sobel(image, cv2.CV_64F, 1, 0, ksize=5)
sobel_y = cv2.Sobel(image, cv2.CV_64F, 0, 1, ksize=5)

final_sobel = cv2.add(sobel_x, sobel_y)

plt.figure(figsize=(10, 7))

plt.subplot(2, 3, 1)
plt.imshow(image, cmap='gray')
plt.title('Original Image')

plt.subplot(2, 3, 2)
plt.imshow(sobel_x, cmap='gray')
plt.title('Sobel X')

plt.subplot(2, 3, 3)
plt.imshow(sobel_y, cmap='gray')
plt.title('Sobel Y')

plt.subplot(2, 3, 4)
plt.imshow(final_sobel, cmap='gray')
plt.title('Final Sobel')

plt.show()
```

```python
image = cv2.imread('images/little_girl.jpg')

sobel_x = cv2.Sobel(image, cv2.CV_64F, 1, 0, ksize=5)
sobel_y = cv2.Sobel(image, cv2.CV_64F, 0, 1, ksize=5)

sobel_x = np.abs(sobel_x)
sobel_y = np.abs(sobel_y)

final_sobel = cv2.add(sobel_x, sobel_y)

plt.figure(figsize=(12, 7))

plt.subplot(2, 3, 1)
plt.imshow(image, cmap='gray')
plt.title('Original Image')

plt.subplot(2, 3, 4)
plt.imshow(sobel_x, cmap='gray')
plt.title('Sobel X')

plt.subplot(2, 3, 5)
plt.imshow(sobel_y, cmap='gray')
plt.title('Sobel Y')

plt.subplot(2, 3, 6)
plt.imshow(final_sobel, cmap='gray')
plt.title('Final Sobel')

plt.show()
```

```python
scr = cv2.imread('images/little_girl.jpg')

scr = cv2.GaussianBlur(scr, (3, 3), 0)

gray = cv2.cvtColor(scr, cv2.COLOR_BGR2GRAY)

gray_x = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)
gray_y = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)

abs_grad_x = cv2.convertScaleAbs(gray_x)
abs_grad_y = cv2.convertScaleAbs(gray_y)

final_gray = cv2.addWeighted(abs_grad_x, 0.5, abs_grad_y, 0.5, 0)

plt.figure(figsize=(12, 7))

plt.subplot(2, 3, 1)
plt.imshow(scr[..., ::1])
plt.title('Original Image')

plt.subplot(2, 3, 4)
plt.imshow(abs_grad_x, cmap='gray')
plt.title('Sobel X')

plt.subplot(2, 3, 5)
plt.imshow(abs_grad_y, cmap='gray')
plt.title('Sobel Y')

plt.subplot(2, 3, 6)
plt.imshow(final_gray, cmap='gray')
plt.title('Final Sobel')

plt.show()
```
