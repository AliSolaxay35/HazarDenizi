```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os
from google.colab import files
```

```python
os.makedirs('images', exist_ok=True)

print("Please Upload The Required Images:")
print("- orange.png")
print("- orange-mask.png")

uploaded = files.upload()

for filename in uploaded.keys():
  dest = os.path.join('images', filename)
  os.rename(filename, dest)
  print(f"Uploaded: {filename}")

print("\nContents Of Images Folder:")
print(os.listdir('images'))
```

```python
scr = cv2.imread('images/orange.png')
mask = cv2.imread('images/orange-mask.png')

results = cv2.illuminationChange(scr, mask)

plt.figure(figsize=(15, 12))

plt.subplot(1, 3, 1)
plt.title('Original Image')
plt.imshow(scr[..., ::-1])

plt.subplot(1, 3, 2)
plt.title('Mask')
plt.imshow(mask[..., ::-1])

plt.subplot(1, 3, 3)
plt.title('Illumination Change')
plt.imshow(results[..., ::-1])

plt.show()
```
