```python
def print_min_and_max(img):
  print("Min: {}, Max: {}, dtype: {}".format(np.min(img), np.max(img), img.dtype))
```

```python
image = cv2.imread('images/gradient.png')

kernel = np.array([[0,0,0],
                   [-1/2,0,1/2],
                   [0,0,0]])

edges = cv2.filter2D(image, -1, kernel)

plt.figure(figsize=(12, 7))

plt.subplot(1, 2, 1)
plt.imshow(image, cmap='gray')
plt.title('Original Image')

plt.subplot(1, 2, 2)
plt.imshow(edges, cmap='gray')
plt.title('Edge Image')

plt.show()

print_min_and_max(image)
print_min_and_max(edges)
```

```python
image = cv2.imread('images/little_girl.jpg', 0)

kernel1 = np.array([[0,0,0],
                   [-1/2,0,1/2],
                   [0,0,0]])

kernel2 = np.array([[0,0,0],
                   [1/2,0,-1/2],
                   [0,0,0]])

edges1 = cv2.filter2D(image, -1, kernel1)
edges2 = cv2.filter2D(image, -1, kernel2)
all_vertical_edges = cv2.add(edges1, edges2)

plt.figure(figsize=(12, 7))

plt.subplot(2, 3, 1)
plt.imshow(image, cmap='gray')
plt.title('Original Image')

plt.subplot(2, 3, 4)
plt.imshow(edges1, cmap='gray')
plt.title('Horizontal Edges')

plt.subplot(2, 3, 5)
plt.imshow(edges2, cmap='gray')
plt.title('Vertical Edges')

plt.subplot(2, 3, 6)
plt.imshow(all_vertical_edges, cmap='gray')
plt.title('Final Edges')

plt.show()
```

```python
image = cv2.imread('images/little_girl.jpg', 0)

kernel = np.array([[0,0,0],
                   [-1/2,0,1/2],
                   [0,0,0]])

edges = cv2.filter2D(image, cv2.CV_64F, kernel)

all_edges = np.abs(edges).astype(np.uint8)

plt.figure(figsize=(12, 7))

plt.subplot(1, 2, 1)
plt.imshow(all_vertical_edges, cmap='gray')
plt.title('Vertical Edges')

plt.subplot(1, 2, 2)
plt.imshow(all_edges, cmap='gray')
plt.title('Final Edges')

plt.show()
```
