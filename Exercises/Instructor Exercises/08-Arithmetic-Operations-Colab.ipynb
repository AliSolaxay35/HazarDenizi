```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
from google.colab import files
```

```python
m1 = np.array([[150, 155], [156, 157]])
m2 = np.ones([2, 2], dtype="uint8") * 100
print("m1:")
print(m1)
print("m2:")
print(m2)
print("m1+m2:")
print(m1 + m2)
```

```python
m1 = np.array([[150, 155], [156, 157]])
m2 = np.ones([2, 2], dtype="uint8") * 100
modified_m1 = m1.astype('int16') + m2
modified_m1 = np.minimum(modified_m1, 255)
modified_m1 = modified_m1.astype('uint8')
modified_m1
```

```python
m1 = np.array([[150, 155], [156, 157]], dtype="uint8")
value_to_add = 100
max_threshold = 255 - 100
m1[m1 >= max_threshold] = 255
m1[m1 < max_threshold] += value_to_add
print(m1)
```

```python
uploaded = files.upload()
image_name_1 = next(iter(uploaded))
input_image_path = '/content/' + image_name_1
print('Selected image 1:', input_image_path)
```

```python
from re import sub
image = cv2.imread(input_image_path)
M = np.ones(image.shape, dtype = "uint8") * 100
added = cv2.add(image, M)
subtracted = cv2.subtract(image, M)

plt.figure(figsize=(15, 5))
plt.subplot(131); plt.imshow(image[..., ::-1]); plt.title('Original')
plt.subplot(132); plt.imshow(added[..., ::-1]); plt.title('Added')
plt.subplot(133); plt.imshow(subtracted[..., ::-1]); plt.title('Subtracted')
plt.show()
```

```python
uploaded = files.upload()
image_name_2 = next(iter(uploaded))
class_vision_image_path = '/content/' + image_name_2
print('Selected image 2:', class_vision_image_path)
```

```python
img2 = cv2.imread(class_vision_image_path)
shape = img2.shape
img1 = cv2.imread(input_image_path)[100:100 + shape[0], 100:100 + shape[1], :]

output_image = cv2.add(img1, img2)

plt.figure(figsize=(10, 5))
plt.subplot(131); plt.imshow(img1[..., ::-1]); plt.title('image1')
plt.subplot(132); plt.imshow(img2[..., ::-1]); plt.title('image2')
plt.subplot(133); plt.imshow(output_image[..., ::-1]); plt.title('image1 + image2')
plt.show()
```

```python
img1 = cv2.imread(input_image_path)
img2 = cv2.imread(class_vision_image_path)

print(img1.shape)
print(img2.shape)
```

```python
logo_h, logo_w, _ = img2.shape
location_h = 10
location_w = 500
padded_im2 = np.zeros_like(img1)
padded_im2[location_h:location_h + logo_h, location_w:location_w + logo_w, :] = img2
plt.imshow(padded_im2[..., ::-1])
plt.show()
```

```python
output_image = cv2.add(img1, padded_im2)
plt.imshow(output_image[..., ::-1])
plt.show()
```

```python
uploaded = files.upload()
image_name_3 = next(iter(uploaded))
wolf_image_path = '/content/' + image_name_3
print('Selected image 3:', wolf_image_path)
```

```python
img3 = cv2.imread(wolf_image_path)
img2 = cv2.imread(class_vision_image_path)
img2_resized = cv2.resize(img2, (img3.shape[1], img3.shape[0]))
output_image = cv2.addWeighted(img3, 0.3, img2_resized, 0.7, 0)

plt.figure(figsize=(10, 5))
plt.subplot(131); plt.imshow(img3[..., ::-1]); plt.title('image3')
plt.subplot(132); plt.imshow(img2_resized[..., ::-1]); plt.title('image2 Resized')
plt.subplot(133); plt.imshow(output_image[..., ::-1]); plt.title('weighted sum')
plt.show()
```

```python
output_image2_8 = cv2.addWeighted(img3, 0.8, img2_resized, 0.2, 0)
outout_image5_5 = cv2.addWeighted(img3, 0.5, img2_resized, 0.5, 0)
output_image8_2 = cv2.addWeighted(img3, 0.2, img2_resized, 0.8, 0)

plt.figure(figsize=(10, 5))
plt.subplot(131); plt.imshow(output_image2_8[..., ::-1]); plt.title('output_image2_8')
plt.subplot(132); plt.imshow(outout_image5_5[..., ::-1]); plt.title('outout_image5_5')
plt.subplot(133); plt.imshow(output_image8_2[..., ::-1]); plt.title('output_image8_2')
plt.show()
```
