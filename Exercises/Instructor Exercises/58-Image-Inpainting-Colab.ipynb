```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os
from google.colab import files
```

```python
os.makedirs('images', exist_ok=True)

print("Please Upload The Required Images:")
print("- Lincoln.jpg")
print("- inpaintMask.png")
print("- input-inpaint.jpg")
print("- input_inpaintMask.png")
print("- blemish.png")
print("- blemishMask.png")

uploaded = files.upload()

for filename in uploaded.keys():
  dest = os.path.join('images', filename)
  os.rename(filename, dest)
  print(f"Uploaded: {filename}")

print("\nContents Of Images Folder:")
print(os.listdir('images'))
```

```python
image = cv2.imread('images/Lincoln.jpg')
mask = cv2.imread('images/inpaintMask.png', cv2.IMREAD_GRAYSCALE)

res_telea = cv2.inpaint(image, mask, 3, cv2.INPAINT_TELEA)
res_ns = cv2.inpaint(image, mask, 3, cv2.INPAINT_NS)

plt.figure(figsize=(18, 7))

plt.subplot(1, 4, 1)
plt.imshow(image[..., ::-1])
plt.title('Original Image')

plt.subplot(1, 4, 2)
plt.imshow(res_telea[..., ::-1])
plt.title('Inpainted Image (TELEA)')

plt.subplot(1, 4, 3)
plt.imshow(res_ns[..., ::-1])
plt.title('Inpainted Image (NS)')

plt.subplot(1, 4, 4)
plt.imshow(mask, cmap='gray')
plt.title('Mask')

plt.show()
```

```python
image = cv2.imread("images/input-inpaint.jpg")
mask = cv2.imread("images/input_inpaintMask.png", 0)

res_telea = cv2.inpaint(image, mask, 3, cv2.INPAINT_TELEA)
res_ns = cv2.inpaint(image, mask, 3, cv2.INPAINT_NS)

plt.figure(figsize=(15, 12))

plt.subplot(2, 2, 1)
plt.imshow(image[..., ::-1])
plt.title('Original Image')

plt.subplot(2, 2, 2)
plt.imshow(res_telea[..., ::-1])
plt.title('Inpainted Image (TELEA)')

plt.subplot(2, 2, 3)
plt.imshow(res_ns[..., ::-1])
plt.title('Inpainted Image (NS)')

plt.subplot(2, 2, 4)
plt.imshow(mask, cmap='gray')
plt.title('Mask')

plt.show()
```
