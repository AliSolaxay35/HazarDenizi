```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os
from google.colab import files
```

```python
os.makedirs('image', exist_ok=True)

print("Please Upload The Required Images:")
print("- triangle.jpg")
print("- clock.jpg")

uploaded = files.upload()

for filename in uploaded.keys():
  dest = os.path.join('image', filename)
  os.rename(filename, dest)
  print(f"File '{filename}' Moved To '{dest}'.")

print("\nContents Of Images Folder:")
print(os.listdir('image'))
```

```python
image = cv2.imread('image/triangle.jpg')
rows, cols, ch = image.shape

pts1 = np.float32([[100, 350],[100, 90],[490, 350]])
pts2 = np.float32([[200, 400],[500, 300],[200, 200]])

M = cv2.getAffineTransform(pts1, pts2)
dst = cv2.warpAffine(image, M, (cols, rows))

plt.subplot(121), plt.imshow(image), plt.title('Input')
plt.subplot(122), plt.imshow(dst), plt.title('Output')

plt.show()
```
