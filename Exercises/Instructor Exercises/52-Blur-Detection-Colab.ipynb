```python
def variance_of_laplacian(image):
  return cv2.Laplacian(image, cv2.CV_64F).var()
```

```python
image1 = cv2.imread('images/plate_cropped.jpg', 0)
image2 = cv2.imread('images/plate3_cropped.jpg', 0)

metric_1 = variance_of_laplacian(image1)
metric_2 = variance_of_laplacian(image2)

plt.figure(figsize=(10, 7))

plt.subplot(1, 2, 1)
plt.imshow(image1, cmap='gray')
plt.title('Image 1')

plt.subplot(1, 2, 2)
plt.imshow(image2, cmap='gray')
plt.title('Image 2')

plt.show()

print(f"Variance Of Laplacian For Image 1: {metric_1}")
print(f"Variance Of Laplacian For Image 2: {metric_2}")
```
