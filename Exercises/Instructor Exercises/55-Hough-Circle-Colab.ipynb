```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os
from google.colab import files
```

```python
os.makedirs('images', exist_ok=True)

print("Please Upload The Required Images:")
print("- smarties.png")
print("- old-iranian-coins.jpg")
print("- iranian-coin.jpg")

uploaded = files.upload()

for filename in uploaded.keys():
  dest = os.path.join('images', filename)
  os.rename(filename, dest)
  print(f"{filename} Uploaded Successfully")

print("\nContents Of Images Folder:")
print(os.listdir('images'))
```

```python
scr = cv2.imread('images/smarties.png')

gray = cv2.cvtColor(scr, cv2.COLOR_BGR2GRAY)
gray = cv2.medianBlur(gray, 5)

rows = gray.shape[0]
circles = cv2.HoughCircles(gray, cv2.HOUGH_GRADIENT, 1, rows / 8,
                           param1=100, param2=30)

if circles is not None:
  circles = np.uint16(np.around(circles))

  for c in circles[0, :]:
    center_x, center_y, radius = c
    center = (center_x, center_y)
    cv2.circle(scr, center, radius, (255, 0, 255), 3)

plt.imshow(scr[..., ::-1])

plt.show()
```

```python
scr = cv2.imread('images/old-iranian-coins.jpg')

gray = cv2.cvtColor(scr, cv2.COLOR_BGR2GRAY)
gray = cv2.medianBlur(gray, 5)

rows = gray.shape[0]
circles = cv2.HoughCircles(gray, cv2.HOUGH_GRADIENT, 1, rows / 8,
                           param1=100, param2=30)

if circles is not None:
  circles = np.uint16(np.around(circles))

  for c in circles[0, :]:
    center_x, center_y, radius = c
    center = (center_x, center_y)
    cv2.circle(scr, center, 1, (0, 100, 100), 3)
    cv2.circle(scr, center, radius, (255, 0, 255), 3)

plt.imshow(scr[..., ::-1])
```
