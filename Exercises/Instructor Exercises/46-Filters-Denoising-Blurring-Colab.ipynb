```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os
from google.colab import files
```

```python
image = cv2.imread('images/noisy-cow.png')

kernel_3x3 = np.ones((3, 3), np.float32) / 9
blurred = cv2.filter2D(image, -1, kernel_3x3)

kernel_7x7 = np.ones((7, 7), np.float32) / 49
blurred2 = cv2.filter2D(image, -1, kernel_7x7)

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(image[..., ::-1])
plt.title('Original Image')

plt.subplot(1, 3, 2)
plt.imshow(blurred[..., ::-1])
plt.title('Blurred Image (3x3 Kernel)')

plt.subplot(1, 3, 3)
plt.imshow(blurred2[..., ::-1])
plt.title('Blurred Image (7x7 Kernel)')

plt.show()
```

```python
image = cv2.imread('images/noisy-cow.png')

blur = cv2.blur(image, (7, 7))

blurred = cv2.GaussianBlur(image, (7, 7), 0)

median = cv2.medianBlur(image, 5)

bilateral = cv2.bilateralFilter(image, 9, 75, 75)

plt.figure(figsize=(15, 10))

plt.subplot(2, 3, 1)
plt.imshow(image[..., ::-1])
plt.title('Original Image')

plt.subplot(2, 3, 2)
plt.imshow(blur[..., ::-1])
plt.title('Blurred Image')

plt.subplot(2, 3, 3)
plt.imshow(blurred[..., ::-1])
plt.title('Gaussian Blurred Image')

plt.subplot(2, 3, 4)
plt.imshow(median[..., ::-1])
plt.title('Median Blurred Image')

plt.subplot(2, 3, 5)
plt.imshow(bilateral[..., ::-1])
plt.title('Bilateral Blurred Image')

plt.show()
```

```python
image = cv2.imread('images/noisy-cow.png')

gaussianBlur2 = cv2.GaussianBlur(image, (11, 11), 2)
gaussianBlur3 = cv2.GaussianBlur(image, (11, 11), 3)
gaussianBlur5 = cv2.GaussianBlur(image, (11, 11), 5)

plt.figure(figsize=(10, 10))

plt.subplot(2, 2, 1)
plt.imshow(image[..., ::-1])
plt.title('Original Image')

plt.subplot(2, 2, 2)
plt.imshow(gaussianBlur2[..., ::-1])
plt.title('Gaussian Blurred Image (sigmaX=2)')

plt.subplot(2, 2, 3)
plt.imshow(gaussianBlur3[..., ::-1])
plt.title('Gaussian Blurred Image (sigmaX=3)')

plt.subplot(2, 2, 4)
plt.imshow(gaussianBlur5[..., ::-1])
plt.title('Gaussian Blurred Image (sigmaX=5)')

plt.show()
```

```python
image = cv2.imread('images/noisy-cow.png')

dst = cv2.fastNlMeansDenoisingColored(image, None, 6, 6, 7, 21)

plt.figure(figsize=(10, 10))

plt.subplot(1, 2, 1)
plt.imshow(image[..., ::-1])
plt.title('Original Image')

plt.subplot(1, 2, 2)
plt.imshow(dst[..., ::-1])
plt.title('Denoised Image')

plt.show()
```
