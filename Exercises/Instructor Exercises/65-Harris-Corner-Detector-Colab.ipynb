```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os
from google.colab import files
```

```python
os.makedirs('images', exist_ok=True)

print("Please Upload The Required Images:")
print("- chessboard.png")
print("- blox.jpg")

uploaded = files.upload()

for filename in uploaded.keys():
  dest = os.path.join('images', filename)
  os.rename(filename, dest)
  print(f"{filename} Has Been Uploaded To {dest}")

print("\nContents Of Images Folder:")
print(os.listdir('images'))
```

```python
img = cv2.imread('images/chessboard.png')

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
gray = np.float32(gray)

dst = cv2.cornerHarris(gray, 2, 3, 0.04)
dst = cv2.dilate(dst, None, iterations=30)

img[dst > 0.01 * dst.max()] = [0, 0, 255]

plt.imshow(img, cmap='gray')
```

```python
img = cv2.imread('images/blox.jpg')

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
gray = np.float32(gray)

corners = cv2.cornerHarris(gray, 2, 3, 0.04)

dst = cv2.dilate(dst, None, iterations=1)

img[corners > 0.01 * corners.max()] = [0, 0, 255]

plt.imshow(img, cmap='gray')
```
