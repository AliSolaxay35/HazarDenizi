```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
canvas = np.ones((600, 800, 3), dtype=np.uint8) * 255
```

```python
house_x1, house_y1 = 220, 280
house_x2, house_y2 = 580, 520

house_color = (180, 220, 255)
outline_color = (0, 0, 0)

cv2.rectangle(canvas, (house_x1, house_y1), (house_x2, house_y2), house_color, -1)
cv2.rectangle(canvas, (house_x1, house_y1), (house_x2, house_y2), outline_color, 5)
```

```python
roof_points = np.array([
    [170, 280],
    [400, 100],
    [630, 280]
], dtype=np.int32)

roof_color = (120, 170, 255)
cv2.fillPoly(canvas, [roof_points], roof_color)
cv2.polylines(canvas, [roof_points], isClosed=True, color=outline_color, thickness=5)
```

```python
cv2.rectangle(canvas, (365, 390), (435, 520), (120, 80, 40), -1)
cv2.rectangle(canvas, (365, 390), (435, 520), outline_color, 5)

cv2.circle(canvas, (420, 445), 7, (0, 255, 255), -1)

cv2.rectangle(canvas, (255, 345), (335, 425), (255, 230, 120), -1)
cv2.rectangle(canvas, (255, 345), (335, 425), outline_color, 5)
cv2.line(canvas, (295, 345), (295, 425), outline_color, 3)
cv2.line(canvas, (255, 385), (335, 385), outline_color, 3)

cv2.rectangle(canvas, (465, 345), (545, 425), (255, 230, 120), -1)
cv2.rectangle(canvas, (465, 345), (545, 425), outline_color, 5)
cv2.line(canvas, (505, 345), (505, 425), outline_color, 3)
cv2.line(canvas, (465, 385), (545, 385), outline_color, 3)
```

```python
chimney_color = (90, 90, 90)
cv2.rectangle(canvas, (485, 150), (540, 260), chimney_color, -1)
cv2.rectangle(canvas, (485, 150), (540, 260), outline_color, 5)
cv2.line(canvas, (475, 150), (550, 150), outline_color, 8)

cv2.circle(canvas, (90, 90), 45, (0, 215, 255), -1)
cv2.circle(canvas, (90, 90), 45, outline_color, 4)

for start, end in [
    ((90, 25), (90, 5)),
    ((90, 155), (90, 175)),
    ((25, 90), (5, 90)),
    ((155, 90), (175, 90)),
    ((45, 45), (30, 30)),
    ((135, 45), (150, 30)),
    ((45, 135), (30, 150)),
    ((135, 135), (150, 150))
]:
     cv2.line(canvas, start, end, (0, 170, 255), 5)

cv2.rectangle(canvas, (55, 390), (90, 520), (40, 90, 140), -1)
cv2.rectangle(canvas, (55, 390), (90, 520), outline_color, 4)

tree_color = (40, 170, 40)
for points in [
    [[15, 400], [72, 270], [130, 400]],
    [[25, 350], [72, 235], [120, 350]],
    [[35, 305], [72, 205], [110, 305]]
]:
    tree_points = np.array(points, dtype=np.int32)
    cv2.fillPoly(canvas, [tree_points], tree_color)
    cv2.polylines(canvas, [tree_points], isClosed=True, color=outline_color, thickness=3)

cv2.line(canvas, (0, 520), (800, 520), (60, 150, 60), 6)
```

```python
canvas_rgb = cv2.cvtColor(canvas, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(10, 8))
plt.imshow(canvas_rgb)
plt.title('A Simple House')
plt.axis('off')
plt.show()
```
