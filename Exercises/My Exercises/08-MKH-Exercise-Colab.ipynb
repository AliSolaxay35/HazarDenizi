```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
```

```python
files.uploaded = files.upload()
input_name = next(iter(files.uploaded))
input_path = '/content/' + input_name

print('Input Image:', input_name)
```

```python
image = cv2.imread(input_path)
if image is None:
  raise ValueError('Image Not Found')

gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

print('Image Shape:', image.shape)
print('Image Size:', image.size)
```

```python
blur_kernel = 51
blur_image = cv2.GaussianBlur(gray_image, (blur_kernel, blur_kernel), 0)

print('Blur Kernel:', blur_kernel)
```

```python
otsu_threshold, binary_image = cv2.threshold(blur_image, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

print('Otsu Threshold:', otsu_threshold)
```

```python
plt.figure(figsize=(16, 5))

plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(blur_image, cmap='gray')
plt.title('Gaussian Blur')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(binary_image, cmap='gray')
plt.title('Binary Image')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
out_path = '/content/Adad_binary.png'

cv2.imwrite(out_path, binary_image)
print('Saved Output:', out_path)

files.download(out_path)
print('Download Output:', out_path)
```
