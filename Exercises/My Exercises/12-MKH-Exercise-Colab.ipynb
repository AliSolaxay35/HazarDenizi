```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os
from google.colab import files
```

```python
uploaded = files.upload()

if len(uploaded) == 0:
  raise Exception("No File Uploaded")

image_name = list(uploaded.keys())[0]
image_path = '/content/' + image_name

img = cv2.imread(image_path)

if img is None:
  raise Exception("Invalid Image")

print('Input Image:', image_name)
print('Image Shape:', img.shape)

plt.figure(figsize=(10, 6))

plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Original Image (Hidden Text)')
plt.axis('off')

plt.show()
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

print('Gray Min/Max:', gray.min(), gray.max())
print('Unique Values:', np.unique(gray))
print('Mean:', gray.mean())
print('STD:', gray.std())

plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.imshow(gray, cmap='gray')
plt.title('Grayscale Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.hist(gray.ravel(), bins=256, range=(0, 255))
plt.title('Histogram Of Grayscale Values')

plt.tight_layout()
plt.show()
```

```python
equalized = cv2.equalizeHist(gray)

print('Equalized Min/Max:', equalized.min(), equalized.max())

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(equalized, cmap='gray')
plt.title('Equalized Grayscale Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.hist(equalized.ravel(), bins=256, range=(0, 256))
plt.title('Histogram Of Equalized Grayscale Values')

plt.tight_layout()
plt.show()
```

```python
clahe = cv2.createCLAHE(clipLimit=5.0, tileGridSize=(8, 8))
clahe_img = clahe.apply(gray)

print('CLAHE Min/Max:', clahe_img.min(), clahe_img.max())

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(clahe_img, cmap='gray')
plt.title('CLAHE Equalized Grayscale Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.hist(clahe_img.ravel(), bins=256, range=(0, 256))
plt.title('Histogram Of CLAHE Equalized Grayscale Values')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(gray, cmap='gray')
plt.title('Grayscale Image')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(equalized, cmap='gray')
plt.title('Equalized Grayscale Image')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(clahe_img, cmap='gray')
plt.title('CLAHE Equalized Grayscale Image')
plt.axis('off')

plt.tight_layout()
plt.show()
```
